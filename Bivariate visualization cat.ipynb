```python
import pandas as pd
import seaborn as sns
```

```python
df=sns.load_dataset("titanic")
df.head()
```

```python
sns.countplot(x="survived",data=df)
```

```python
sns.countplot(x="survived",data=df,hue="class")
```

```python
## when we've more no. of values in a prticular category then will use heat map
```

```python
## Heat map
## easier to study and will use when we've more values in our cat data
x=pd.crosstab(df["survived"],df["class"])
x
```

```python
sns.heatmap(x,annot=True,fmt=".1f")
## it is very useful when we've to analyze 2 cat
```

```python
sns.stripplot(x="class",y="age",hue="sex",data=df)
```

```python
df2=sns.load_dataset("iris")
df2.head(10).plot(kind="bar")
```

```python
df2.head(10).plot(kind="bar",stacked="True")
## if we've more cat in one bar then will give stacked value
```

```python
## when we've numerical data in our dataset then will use pair plot
sns.pairplot(df2)
```
