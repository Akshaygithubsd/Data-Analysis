Graphical analysis can be divided into 2 parts , univariate and bivariate analysis.
in univariate we study particular column or a field or a variable. in bivariate we discuss relationship b/w 2 columns. and in multivariate will discuss relationship more than 2 columns.


```python
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv("AB_NYC_2019.csv")
df.head(2)
```

## Categorical

## Bargraph

```python
sns.countplot(data=df,x="neighbourhood_group")
```

```python
sns.countplot(data=df,x="neighbourhood_group",hue="room_type")
```

## Pie Chart

```python
df["room_type"].value_counts()
```

```python
df["room_type"].value_counts().plot(kind="pie")
```

```python
df["room_type"].value_counts().plot(kind="pie", autopct="%.2f")
```

```python
df["room_type"].value_counts().plot(kind="bar")
```
